## Lists

### List Slicing

```python
# List slicing examples: [start:stop:step]
myList = [1,2,3,4,5]
print(f'myList[3:]={myList[3:]}')
print(f'myList[0:6:2]={myList[0:6:2]}')
print(f'myList[0:6:3]={myList[0:6:3]}')
print(f'myList[::2]={myList[::2]}')
```

```python
# Custom list subclass that formats output in rows of 10
class Lista(list):
    # Override the __repr__ method to format list output in rows of 10
    def __repr__ (self):
        aux = "["
        index = 0
        for i in self:
            # Insert a newline every 10 items for readability
            if index % 10 == 0 and index > 0:
                aux += "\n"
            aux += str(i)  
            # Add a comma after every item except the last one
            if i < len(self) - 1:
                aux += ","
            index += 1
        aux += "]"
        return aux

x = Lista()
for i in range(100):
    x.append(i)
print(x)
```

```python
# Demonstrate slicing with a step and reverse slicing
myList = list(range(100))
step_ten=myList[::10] #array[start:stop:step] 
step_reverse_ten=myList[::-10]
print(step_ten)
print(step_reverse_ten)
```

### Modifying Lists

```python
# Examples of list modification methods: append, insert, remove, pop
myList = [1,2,3,4]
myList.append(5)
print(myList)
myList.insert(3, 'a new value')
print(myList)
myList.remove('a new value')
print(myList)
print("Elemento eliminado: ",myList.pop())
print(myList)
while len(myList):
    print("Elemento eliminado: ",myList.pop())
print(myList)
```

```python
# Demonstrates that assignment creates a shared reference to the same list
# Modifying `a` will also be visible through `b`
# Shared References
a = [1,2,3,4,5]
b = a
a.append(6)
print(b)
```

```python
# Copying a list to create an independent shallow copy
a = [1,2,3,4,5]
b = a.copy()
a.append(6)
print(a)
print(b)
```
