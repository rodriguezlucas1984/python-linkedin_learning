## Sets

```python
# Create a set and show its type and contents
mySet = {'a', 'b', 'c'}
print(f'{type(mySet)}: {mySet}')

# Another way to create a set from an iterable
mySet = set(('a', 'b', 'c'))
print(f'{type(mySet)}: {mySet}')

# Create a list with duplicates and show it
myList = ['a', 'b', 'b', 'c', 'c']
print(f'{type(myList)} {myList}')

# Convert the list to a set to remove duplicates, then back to a list
myList = list(set(myList))
print(f'{type(myList)} {myList}')
```

```python
# Sets are not subscriptable (they do not support indexing)
# Attempting to access by index raises an error
try:
    print(mySet[0])
except Exception as e:
    print(e)
```

```python
# Add a new element to the set (sets ignore duplicates)
mySet.add('d')
print(mySet)
```

```python
# Membership testing: check if an element is present in the set
print(f"'a' in mySet:{'a' in mySet}")
print(f"'z' in mySet:{'z' in mySet}")
```

```python
# Remove items from the set one by one using pop()
# Note: pop() removes and returns an arbitrary element
len(mySet)
while len(mySet):
    print(mySet.pop())
```

```python
# Display the current state of mySet
mySet
```

```python
# Recreate the set for further examples
mySet = {'a', 'b', 'c'}
```

```python
# Remove an element safely using discard (no error if missing)
mySet.discard('a')
```

```python
# Show the set after discard
mySet
```

## Tuples

```python
# Define a tuple and display it
myTuple = ('a', 'b', 'c')
print(myTuple)
print("Item 0: ",myTuple[0])
```

```python
# Tuples are immutable - attempting to assign to an index raises a TypeError
try:
    myTuple[0] = 'd'
except TypeError as te:
    print(te)
```

```python
# Function that returns multiple values as a tuple
def returnsMultipleValues():
    return 1,2,3

type(returnsMultipleValues())
```

```python
# Create a numeric tuple and check its type
myTuple = (1,2,3)
type(myTuple)
```

```python
# Unpack the returned tuple into variables and print each value
a, b, c = returnsMultipleValues()
print(a)
print(b)
print(c)
```
