## List Comprehensions

```python
# Create a list of numbers and use a list comprehension to double each value.
myList = [1,2,3,4,5]
[2*item for item in myList]
```

### List comprehensions with filters

```python
# Create a list of numbers from 0 to 99.
myList = list(range(100))
# Keep only numbers divisible by 10.
filteredList = [item for item in myList if item % 10 == 0]
print(filteredList)
```

```python
# Keep only numbers whose last digit is 0, 1, or 2.
filteredList = [item for item in myList if item % 10 < 3]
print(filteredList)
```

### List comprehensions with functions

```python
# Use a string with sentences to demonstrate list comprehensions on words.
myString = 'My name is Ryan Mitchell. I live in Boston'
print(myString.split('.'))
```

```python
# Split the string into words using whitespace.
myString.split()
```

```python
# Define a helper function that removes punctuation and converts words to lowercase.
def cleanWord(word):
    return word.replace('.', '').lower()

# Apply the cleaning function to every word in the string.
[cleanWord(word) for word in myString.split()]
```

```python
# Use a filter to keep only short cleaned words.
[cleanWord(word) for word in myString.split() if len(cleanWord(word)) < 3]
```

### Nested list comprehensions

```python
# Use nested comprehensions to clean words in each sentence separately.
[[cleanWord(word) for word in sentence.split()] for sentence in myString.split('.')]
```
