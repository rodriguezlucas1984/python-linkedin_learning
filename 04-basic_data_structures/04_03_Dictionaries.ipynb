```python
from collections import defaultdict
```

## Dictionaries

```python
# Create a simple dictionary mapping letters to animal names.
animals = {
    'a': 'aardvark',
    'b': 'bear',
    'c': 'cat',
}
print(animals)
print(animals['a'])
# Add a new key/value pair to the dictionary.
animals['d'] = 'dog'
print(animals)
```

```python
# Update the value for an existing key.
animals['a'] = 'antelope'
animals
```

```python
# Show the dictionary keys view.
animals.keys()
```

```python
# Show the dictionary values view.
animals.values()
```

```python
# Convert the dictionary keys view into a list.
list(animals.keys())
```

```python
# Accessing a missing key directly raises a KeyError.
try:
    animals['e']
except KeyError:
    print("Key 'e' not found.")
```

```python
# Safely retrieve a value with get() to avoid an exception.
animals.get('a')
```

```python
# Get the number of items stored in the dictionary.
len(animals)
```

```python
# Store lists of animals under each letter key.
animals = {
    'a': ['aardvark', 'antelope'],
    'b': ['bear'],
}
```

```python
# Add another animal name to the list under key 'b'.
animals['b'].append('bison')
```

```python
# Create a new list for a key that does not exist yet.
animals['c'] = ['cat']
```

```python
# Ensure the key exists before appending to its list.
if 'c' not in animals:
    animals['c'] = []
    
animals['c'].append('cat')
```

### The Default Dict

```python
# Create a defaultdict that automatically creates empty lists for missing keys.
animals = defaultdict(list)
```

```python
# Inspect the newly created defaultdict.
animals
```

```python
# Append to a list for a missing key; defaultdict creates the list automatically.
animals['e'].append('elephant')
animals
```

```python
# Append another value to the same automatically created list.
animals['e'].append('emu')
animals
```

```python
# Accessing a missing key returns a new empty list instead of raising KeyError.
animals['f']
```
