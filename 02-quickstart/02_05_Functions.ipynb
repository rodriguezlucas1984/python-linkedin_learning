```python
#Factorial
def factorial (x):
    if  not isinstance(x, int) or x<0:
        raise TypeError('x debe ser un entero positivo')
    if x==0:
        return 1;
    return x*factorial(x-1)

y=6
try :
    print(f'{y}! = {factorial(y)}')
except TypeError as msg_error:
    print(msg_error)
```
