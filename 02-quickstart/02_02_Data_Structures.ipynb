# Data Structures

## Lists

```python
# Lista de numeros
my_list = [4,-5,7,0]
print(my_list)
```

```python
# Lista de cadenas de caracteres
my_list = ['list', 'of', 'strings']
print('Lista: ', my_list)
```

```python
# Lista mixta
my_list = [1,'list', False, []]
# Función mostar lista
def mostrarLista(lista):
    for indice,item in enumerate(lista):
        print(f'Item {indice}: {item} (es {type(item)})')

mostrarLista(my_list)
```

```python
my_list = [[1,2,3],[False, True], []]
mostrarLista(my_list)
```

```python
len(my_list)
```

## Sets

```python
# Lista sin elementos repetidos
my_set = {1,2,3,4,5}
#print(my_set)
mostrarLista(my_set)
```

```python
type(my_set)
```

```python
print('Su cantidad de elementos es: ',len(my_set))
# Adición de un elemento existente
my_set.add(5)
print('Su cantidad de elementos es: ',len(my_set))
# Adición de otro elemento
my_set.add(8)
print('Su cantidad de elementos es: ',len(my_set))
```

```python
my_set = {1,1,2,2}
len(my_set)
print(my_set)
```

```python
[1,2] == [2,1]
```

```python
# No importa el orden, solo que tenga los mismos elementos (no adiciona repetidos)
{1,2,3} == {3,2,1,1,1}
```

## Dictionaries

```python
my_dictionary = {
    'manzana': 'Una fruta roja', 
    'oso': 'Un animal aterrador'
}
```

```python
my_dictionary = {
    'manzana': 'Una fruta roja', 
    'oso': 'Un animal aterrador',
    'manzana': 'Aveces son verdes'
}
```

```python
my_dictionary['manzana']
```
