## Arithmetic Operators 

```python
# Suma
a=1
b=3
print(f'{a} + {b} = {a+b}')
```

```python
# Multiplicación
a=4
b=5
print(f'{a} * {b} = {a*b}')
```

```python
# Potencia
a=5
b=3
print(f'{a} ** {b} = {a**b}')
```

```python
# División
a=20
b=5
print(f'{a} / {b} = {a/b}')
print('El resultado es: ', type(a/b))
```

```python
# División
a=20
b=6
print(f'{a} / {b} = {a/b}')
```

```python
# Modulo
a=20
b=6
print(f'{a} % {b} = {a%b}')
```

### Arithmetic Operators with Strings 

```python
#Concatenación
'string 1 ' + 'string 2'
```

```python
# Multiplicación
'- string 1 - ' * 4
```

```python
# Concatenar str con otro tipo de dato dispara un erro de tipo
a='1'
b=4
try:
    a+b
except TypeError as msg_error:
    print(f'la operación \'{a}\' + {b} arroja el siguiente error: \'{msg_error}\'')
else:
    print('No sueñes... esta linea de codigo no se ejecutara')
```

### Logical Operators (and, or, not)

```python
# Operador and
print(f' True   and     True    => {True and True}')
print(f' False  and     True    => {False and True}')
print(f' True   and     False   => {True and False}')
print(f' False  and     False   => {False and False}')
```

```python
# Operador or
print(f' True   or    True    => {True or True}')
print(f' False  or    True    => {False or True}')
print(f' True   or    False   => {True or False}')
print(f' False  or    False   => {False or False}')
```

```python
not True
```

```python
not False
```
