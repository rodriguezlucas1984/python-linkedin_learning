```python
from decimal import Decimal, getcontext
```

## Integers

```python
print(f"int('100')={int('100')}")
#'100' en base 2
print(f"int('100',2)={int('100',2)}")
try:
    print(f"int(100,2)={int(100,2)}")
except TypeError as e:
    print(f"int(100,2)=>{e}")
# '1ab' en base 16
print(f"int('1ab',16)={int('1ab',16)}")
#Error de redondeo
print(f"1.2 - 1.0={1.2 - 1.0}")
```

## Decimals

```python
#Contexto aritmetico actual
print(getcontext())
#Modificación de presición
getcontext().prec=4
print(getcontext())
print(f'Decimal(1)/Decimal(3)={Decimal(1) / Decimal(3)}')
#Modificación de presición
getcontext().prec=2
print(f'Decimal(1)/Decimal(3)={Decimal(1) / Decimal(3)}')
print(f'Decimal(3.14)={Decimal(3.14)}')
print(f"Decimal('3.14')={Decimal('3.14')}")
#Redondeo
print(f'round(1.2 - 1.0, 2)={round(1.2 - 1.0, 2)}')
```
