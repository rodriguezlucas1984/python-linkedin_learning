## Strings

### Slicing

```python
name = 'My name is Ryan Mitchell'
myList = [1,2,3,4,5]
print(f"name= '{name}'","(" + str(type(name))+")")
print(f"myList = {myList}", "(" + str(type(myList))+")")
print(f"name[0]= '{name[0]}'")
print(f"name[1]= '{name[1]}'")
print(f"name[0:7]= '{name[0:7]}'")
print(f"name[:7]= '{name[:7]}'")
print(f"name[11:]= '{name[11:]}'")
print(f"len(name)= {len(name)}")
print(f"len(myList)= {len(myList)}")
print(f"myList[2:4]= {myList[2:4]}")
```

### Formatting

```python
print('My number is: '+str(5))
print(f'My number is: {5}')
print(f'My number is: {5} and twice that is {2*5}')
print(f'Pi is: {math.pi:.2f}')
print('Pi is: {} {}'.format(math.pi, type(math.pi)))
```
