## Ints and Floats

```python
print(f'división:      20 / 4 ={20/4}; {type(20/4)}')
print(f'suma:          4+4.0 ={4+4.0}; {type(4+4.0)}')
print(f'potenciación:  20 / 4 ={20/4}; {type(20/4)}')
print(f'casteo:        \n\tint(4 ** 4.0) ={int(4 ** 4.0)}; {type(int(4 ** 4.0))} ')
print(f'\tint(8.9) ={int(8.9)}; {type(int(8.9))} ')
print(f'\tint(8.99999999) ={int(8.99999999)}; {type(int(8.99999999))} ')
print(f'\tint(14/3) ={int(14/3)}; {type(int(14/3))} ')
print(f'redondeo:        \n\tround(14/3, 2) ={round(14/3, 2)}; {type(round(14/3, 2))} ')
print(f'errores punto flotante:        \n\t1.2 - 1.0 ={1.2 - 1.0}; {type(1.2 - 1.0)} ')
```
