## **Convert hexadecimal to decimal**

Hexadecimal, or "base 16," uses all of the numbers 0–9, plus a few others to signify higher numbers:

* **A = 10**
* **B = 11**
* **C = 12**
* **D = 13**
* **E = 14**
* **F = 15**

For more information about (and some background on) the hexadecimal numbering system, you can watch [this video](https://example.com). (Nota: El enlace en la imagen apunta a un video informativo).

### **Practice with hexadecimal**

The hexadecimal number **2A** represents the decimal number **42**. Why?

The hexadecimal number **2A** contains a **2** in the *16's place*, plus an **A** in the *1's place*.

Think about the decimal number **42**—it contains a **4** in the *10's place* and a **2** in the *1's place*.

Just like decimal has a "10's place," hexadecimal has a corresponding "16's place."

So, we can convert the hexadecimal **2A** into decimal using:


$$(2 * 16) + (A * 1) = (2 * 16) + (10 * 1) = 42$$

Similarly, the hexadecimal number **CAB** contains a **C** (or 12) in the *256's place*, an **A** (or 10) in the *16's place*, and a **B** (or 11) in the *1's place*.

Just like decimal has a "100's place" (which is $10 * 10$), hexadecimal has a "256's place" (which is $16 * 16$).

So we can convert the hexadecimal **CAB** into decimal using:


$$(C * 256) + (A * 16) + (B * 1) = (12 * 256) + (10 * 16) + (11 * 1) = 3243$$

```python
# Converts a string hexadecimal number into an integer decimal
# If hexNum is not a valid hexadecimal number, returns None
def hexToDec(hexNum):
    index=len(hexNum)-1
    exponente=0
    decimal=0
    while index>=0:
        if not hexNum[index] in hexNumbers:
            return None
        decimal= decimal+hexNumbers[hexNum[index]]*(16**exponente)
        exponente=exponente+1
        index=index-1
    return decimal
```
