## Booleans

### Casting Booleans

```python
print(f"bool(1) = {bool(1)}")
print(f"bool(0) = {bool(0)}")
print(f"bool(-1) = {bool(-1)}")
print(f"bool(1j) = {bool(1j)}")
print(f"bool(0.0) = {bool(0.0)}")
print(f"bool(0j) = {bool(0j)}")
print(f"bool('True') = {bool('True')}")
print(f"bool('False') = {bool('False')}")
print(f"bool('') = {bool('')}")
print(f"bool({'{}'}) = {bool({})}")
print(f"bool(None) = {bool(None)}")
```

```python
myList = [1,2]
if bool(myList):
    print('Mylist has some values in it!')
```

```python
a = 5
b = 5
if a - b:
    print('a and b are not equal!')
```

```python
a == b
```

### Boolean Logic

```python
weatherIsNice = True
haveUmbrella = False

if not (haveUmbrella or weatherIsNice):
    print('Stay inside')
else:
    print('Go for a walk')
```

```python
weatherIsNice = True
haveUmbrella = False

if (not haveUmbrella) and (not weatherIsNice):
    print('Stay inside')
else:
    print('Go for a walk')
```

```python
weatherIsNice = True
haveUmbrella = False

if haveUmbrella or weatherIsNice:
    print('Go for a walk')
else:
    print('Stay inside')
```
